```python
import os
import pandas as pd
#การสร้าง ตัวแปร เพื่อเก็บ "ที่อยู่ (Path)" ของโฟลเดอร์ใน Google Drive  (เดือน 7 ถึง 12 ปี 2025)
folder_path = '/content/drive/MyDrive/midtermbsc_2025/Input - (Operation Daily rain from API_ 2025)/2025-07/'
folder_path2 = '/content/drive/MyDrive/midtermbsc_2025/Input - (Operation Daily rain from API_ 2025)/2025-08/'
folder_path3 = '/content/drive/MyDrive/midtermbsc_2025/Input - (Operation Daily rain from API_ 2025)/2025-09/'
folder_path4 = '/content/drive/MyDrive/midtermbsc_2025/Input - (Operation Daily rain from API_ 2025)/2025-10/'
folder_path5 = '/content/drive/MyDrive/midtermbsc_2025/Input - (Operation Daily rain from API_ 2025)/2025-11/'
folder_path6 = '/content/drive/MyDrive/midtermbsc_2025/Input - (Operation Daily rain from API_ 2025)/2025-12/'

#นำตัวแปรที่เก็บ Path ทั้งหมดมารวมกันไว้ใน List เพื่อให้ง่ายต่อการสั่งให้โปรแกรมทำงานวนซ้ำ (Loop) ทีเดียวทุกโฟลเดอร์
all_folder_paths = [
    folder_path,
    folder_path2,
    folder_path3,
    folder_path4,
    folder_path5,
    folder_path6
]
#สร้างลิสต์ว่างๆ ไว้รอเก็บ "แผ่นข้อมูล (DataFrame)" ที่อ่านได้จากแต่ละไฟล์
df_list = []
#สร้างตัวนับจำนวนไฟล์ เริ่มต้นที่ 0 เพื่อเอาไว้สรุปตอนท้ายว่าอ่านไปได้กี่ไฟล์แล้ว
file_count = 0
#การใช้ Loop เพื่ออ่านไฟล์ CSV ทุกไฟล์
#เริ่มเข้าไปดูในแต่ละโฟลเดอร์ตามลิสต์ที่เราทำไว้
#สั่งให้โปรแกรมดูว่าในโฟลเดอร์นั้นมีไฟล์อะไรบ้าง
#เลือกเฉพาะไฟล์ที่นามสกุลลงท้ายด้วย .csv เท่านั้น
for path in all_folder_paths:
    csv_files = [f for f in os.listdir(path) if f.endswith('.csv')]
    for csv_file in csv_files:
        file_path = os.path.join(path, csv_file)
        df = pd.read_csv(file_path)
        df_list.append(df)
        file_count += 1

combined_input_df = pd.concat(df_list, ignore_index=True)

print(f"Successfully combined {file_count} CSV files from all specified folders into one DataFrame.")
display(combined_input_df.tail())
```

```python
df = filtered_combined_df.copy() # Corrected variable name
df['measure_datetime'] = pd.to_datetime(df['measure_datetime'])

dates = pd.date_range(df['measure_datetime'].min(),
                      df['measure_datetime'].max())

full = pd.MultiIndex.from_product(
    [df['station_code'].unique(), dates], # Use 'station_code' as 'tele_station_oldcode' was renamed
    names=['station_code', 'measure_datetime'] # Use 'station_code'
).to_frame(index=False)

missing = (full.merge(df[['station_code','measure_datetime','rainfall_value']], # Use 'station_code'
                       how='left')
                .assign(missing=lambda x: x['rainfall_value'].isna())
                .groupby('station_code')['missing'] # Use 'station_code'
                .agg(missing_count='sum'))

result = (missing
          .assign(total_expected=len(dates),
                  missing_proportion=lambda x: x['missing_count']/len(dates))
          .sort_values('missing_proportion', ascending=False))

result
```

## ข้อ4 ลบสถานีที่ข้อมูลหายเกิน0.2

```python
# Identify stations to be removed (missing_proportion > 0.2)
removed_stations_df = result[result['missing_proportion'] > 0.2]

# Identify stations to be kept (missing_proportion <= 0.2)
remaining_stations_df = result[result['missing_proportion'] <= 0.2]

print("--- รายชื่อสถานีที่ถูกคัดออก (Missing Proportion > 20%) ---")
display(removed_stations_df)

```

##ข้อ5

```python
import pandas as pd

## =========================================================
## STEP 1: เตรียมข้อมูลรายวัน
## =========================================================
df = Dailyrain.copy()

df['measure_datetime'] = pd.to_datetime(df['measure_datetime'])

## สร้าง Day of Year (DOY)
df['doy'] = df['measure_datetime'].dt.dayofyear


## =========================================================
## STEP 2: ตรวจชื่อคอลัมน์ฝน
## =========================================================
RAIN_COL = 'data'   # <<<< แก้ตรงนี้ให้ตรงกับ df.columns (เดิมคือ 'rain')


## =========================================================
## STEP 3: คำนวณ Median ฝนย้อนหลัง
## แยกตามสถานี + DOY
## =========================================================
median_doy = (
    df.groupby(['station_code', 'doy'])[RAIN_COL]
      .median()
      .reset_index()
      .rename(columns={RAIN_COL: 'rainfall_median'})
)


## =========================================================
## STEP 4: สร้าง Full Grid
## =========================================================
dates = pd.date_range(
    df['measure_datetime'].min(),
    df['measure_datetime'].max()
)

full_grid = pd.MultiIndex.from_product(
    [df['station_code'].unique(), dates],
    names=['station_code', 'measure_datetime']
).to_frame(index=False)

full_grid['doy'] = full_grid['measure_datetime'].dt.dayofyear


## =========================================================
## STEP 5: merge + เติมค่า median
## =========================================================
df_imputed = (
    full_grid
    .merge(
        df[['station_code', 'measure_datetime', RAIN_COL]],
        how='left'
    )
    .merge(
        median_doy,
        on=['station_code', 'doy'],
        how='left'
    )
)

df_imputed[RAIN_COL] = (
    df_imputed[RAIN_COL]
    .fillna(df_imputed['rainfall_median'])
)

df_imputed = df_imputed.drop(columns='rainfall_median')


## =========================================================
## STEP 6: Output
## =========================================================
daily_rainfall_doy_median_filled = df_imputed

display(daily_rainfall_doy_median_filled.head())
```

##ข้อ 6

เหตุผลทางสถิติในการเลือกใช้ค่ามัธยฐาน (Median) แทนค่าเฉลี่ย (Mean) ในการวิเคราะห์ปริมาณน้ำฝน
ในการวิเคราะห์ข้อมูลทางด้านอุตุนิยมวิทยา โดยเฉพาะปริมาณน้ำฝน (Rainfall Data) การเลือกใช้ ค่ามัธยฐาน (Median) เป็นตัวแทนของข้อมูลมีความเหมาะสมมากกว่าการใช้ค่าเฉลี่ยเลขคณิต (Mean) ด้วยเหตุผลหลักดังนี้:

1. ความทนทานต่อค่าสุดโต่ง (Robustness to Outliers) ข้อมูลปริมาณน้ำฝนมักปรากฏค่าที่สูงผิดปกติ (Extreme Events) เช่น สภาวะฝนตกหนักจากพายุหรือมรสุม ซึ่งค่าเหล่านี้จะมีปริมาณสูงกว่าปกติหลายเท่าตัว หากใช้ค่าเฉลี่ยเลขคณิต ข้อมูลที่สูงผิดปกติเพียงไม่กี่ค่าจะดึงค่าเฉลี่ยให้สูงขึ้นเกินความเป็นจริง ส่งผลให้การพยากรณ์หรือการตั้งค่าปกติ (Normal Range) คลาดเคลื่อนได้ ในขณะที่ค่ามัธยฐานจะไม่ได้รับผลกระทบจากค่าสุดโต่งเหล่านี้ ทำให้สะท้อนภาพรวมของเหตุการณ์ที่เกิดขึ้นส่วนใหญ่ได้อย่างแม่นยำกว่า

2. ลักษณะการกระจายตัวของข้อมูลแบบไม่สมมาตร (Non-normal Distribution) โดยธรรมชาติ ข้อมูลปริมาณน้ำฝนจะมีการกระจายตัวแบบเบ้ขวา (Right-skewed Distribution) เนื่องจากมีจำนวนวันที่ฝนไม่ตกหรือตกน้อยเป็นจำนวนมาก และมีเพียงไม่กี่วันที่มีฝนตกหนักมาก ในทางสถิติ เมื่อข้อมูลมีการกระจายตัวไม่เป็นรูปโค้งปกติ (Non-Gaussian distribution) ค่าเฉลี่ยจะถูกดึงไปทางทิศทางของหางข้อมูล (Skew) ทำให้ค่ามัธยฐานซึ่งเป็นค่าที่อยู่ตำแหน่งกึ่งกลางของข้อมูลทั้งหมด เป็นตัวแทนที่มีเสถียรภาพมากกว่าในการระบุ "ค่าปกติ" ของพื้นที่นั้นๆ

3. การสะท้อนความเป็นไปได้จริงในเชิงพื้นที่ (Representativeness) การใช้ค่ามัธยฐานช่วยให้ผู้บริหารจัดการน้ำหรือนักวิเคราะห์มองเห็น "ค่ากลางที่เกิดขึ้นจริง" ของสถานีนั้นๆ ได้ดีกว่าค่าเฉลี่ย ซึ่งมักเป็นค่าทางตัวเลขที่ไม่เคยเกิดขึ้นจริงในสถานการณ์ปกติ การใช้ค่ามัธยฐานจึงช่วยลดความเสี่ยงในการประเมินทรัพยากรน้ำเกินจริง (Overestimation) ส่งผลให้การวางแผนรับมือภัยแล้งหรือการจัดสรรน้ำมีประสิทธิภาพและสอดคล้องกับสภาพข้อเท็จจริงในเชิงพื้นที่มากยิ่งขึ้น

```python
import pandas as pd

## =========================================================
## STEP 1: เตรียมข้อมูลฝนย้อนหลัง
## =========================================================
hist = Dailyrain.copy()

hist['measure_datetime'] = pd.to_datetime(hist['measure_datetime'])
# Rename 'data' column to 'rainfall_value' to match expectation
hist.rename(columns={'data': 'rainfall_value'}, inplace=True)

## แยก year / month
hist['year'] = hist['measure_datetime'].dt.year
hist['month'] = hist['measure_datetime'].dt.month


## =========================================================
## STEP 2: ใช้เฉพาะข้อมูลย้อนหลัง (ไม่เอา 2025)
## =========================================================
hist_past = hist[hist['year'] < 2025]


## =========================================================
## STEP 3: กรองเฉพาะสถานีที่เหลืออยู่
## =========================================================
valid_stations = remaining_stations_df.index.tolist()

hist_past_filtered = hist_past[
    hist_past['station_code'].isin(valid_stations)
]


## =========================================================
## STEP 4: กรองเฉพาะเดือน 7–12
## =========================================================
hist_past_filtered = hist_past_filtered[
    hist_past_filtered['month'].between(7, 12)
]


## =========================================================
## STEP 5: คำนวณผลรวมปริมาณน้ำฝนรายเดือน
## =========================================================
monthly_sum = (
    hist_past_filtered
    .groupby(['station_code', 'month'], as_index=False)
    ['rainfall_value']
    .sum()
)


## =========================================================
## STEP 6: ใส่ปี 2025 (label)
## =========================================================
monthly_sum['year'] = 2025


## =========================================================
## STEP 7: จัดรูปแบบ Output
## =========================================================
output_monthly_rainfall = (
    monthly_sum
    .rename(columns={
        'station_code': 'Node',
        'rainfall_value': 'rain'
    })
    [['Node', 'rain', 'month', 'year']]
    .sort_values(['Node', 'month'])
    .reset_index(drop=True)
)


## =========================================================
## STEP 8: ตรวจสอบผลลัพธ์
## =========================================================
print("จำนวนสถานีที่ใช้:", output_monthly_rainfall['Node'].nunique())
print("จำนวนแถวทั้งหมด:", len(output_monthly_rainfall))

output_monthly_rainfall
```
